```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
citas = pd.read_csv('citas.csv')
pacientes = pd.read_csv('pacientes.csv')
medicos = pd.read_csv('medicos.csv')

print("Datos cargados correctamente")
```

```python
# KPI 1

estado_counts = citas['Estado'].value_counts()

porcentaje_completadas = (
    estado_counts.get('Completada', 0) / len(citas)
) * 100

print(f"Porcentaje de citas completadas: {porcentaje_completadas:.2f}%")

plt.figure(figsize=(8,5))

estado_counts.plot(kind='bar')

plt.title('Distribución de Estados de Citas')
plt.xlabel('Estado')
plt.ylabel('Cantidad')

plt.show()
```

```python
# KPI 2

especialidades = citas.groupby('Especialidad').size()

especialidades = especialidades.sort_values(ascending=False)

print(especialidades)

plt.figure(figsize=(10,5))

especialidades.plot(kind='bar')

plt.title('Cantidad de Citas por Especialidad')
plt.xlabel('Especialidad')
plt.ylabel('Cantidad')

plt.xticks(rotation=45)

plt.show()
```

```python
# KPI 3

citas['FechaAgendado'] = pd.to_datetime(citas['FechaAgendado'])

citas['Mes'] = citas['FechaAgendado'].dt.to_period('M')

citas_mes = citas.groupby('Mes').size()

print(citas_mes)

plt.figure(figsize=(10,5))

citas_mes.plot(kind='line', marker='o')

plt.title('Tendencia Mensual de Citas')
plt.xlabel('Mes')
plt.ylabel('Cantidad')

plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(7,7))

estado_counts.plot(
    kind='pie',
    autopct='%1.1f%%'
)

plt.title('Distribución de Estados')
plt.ylabel('')

plt.show()
```
